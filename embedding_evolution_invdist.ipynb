# Embedding Evolution Across Cosmic Time — fiducial invdist embedding

**Copy of `embedding_evolution.ipynb` (2026-09-21) adapted to the new fiducial embeddings** (`snapshot*_invdist_r2_all.npz`: capped 1/d profiles, cap=10, squared kernel = PCA). The legacy notebook is kept unchanged; figures from THIS notebook get an `_invdist` suffix so the legacy figures are not overwritten.

Track the evolution of the embedding (Dims 1, 2, 3) across redshift using the SubLink merger trees. All earlier-snapshot profiles are projected into z=0's coordinate system — with the squared kernel this is an **exact linear projection** through z=0's stored PCA frame (`profile_mean`, `components`), replacing the legacy Nyström machinery.

**Orientation**: the fiducial embedding's native orientation is kept — **high Dim 1 = dense** (corr with n_neighbors = +0.74 at z=0), opposite to the legacy embedding.

**Snapshots used:** 99 (z=0), 50 (z≈1), 33 (z≈2), 25 (z≈3), 21 (z≈4), 17 (z≈5).

**Data:** the full-sample `_all` invdist embeddings and the matching `progenitors_multi_all.csv` / `progenitor_properties_all.csv` (still valid: `original_index` is identical to the legacy files).


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics.pairwise import euclidean_distances
from scipy.sparse.linalg import eigsh

SNAPS = [17, 21, 25, 33, 50, 99]
# Nominal integer redshifts; exact catalog values are
# 99: 0.0000, 50: 0.9973, 33: 2.0020, 25: 3.0081, 21: 4.0079, 17: 4.9959
REDSHIFTS = {99: 0.0, 50: 1.0, 33: 2.0, 25: 3.0, 21: 4.0, 17: 5.0}
H = 0.6774  # TNG cosmology

# Fixed colour per redshift -- sequential blue->red ramp, shared by all
# per-redshift line plots (same scheme as the composite grid figure).
ZCOLORS = {
    5.0: '#ff8f86',   # pale salmon
    4.0: '#ff5650',
    3.0: '#ea082a',   # red
    2.0: '#6c22e8',
    1.0: '#0202da',   # blue
    0.0: '#06068d',   # dark navy
}

def zcolor(z):
    """Colour for a redshift. Falls back to the nearest tabulated z, so exact catalog
    values (0.9973, 3.0081, ...) work as well as the rounded ones."""
    z = float(z)
    if z in ZCOLORS:
        return ZCOLORS[z]
    return ZCOLORS[min(ZCOLORS, key=lambda k: abs(k - z))]

# Exclude per-snapshot zero-neighbor galaxies (all-zero sorted vectors that all
# project to ONE degenerate point; 52% of `tracked` at z=5) from the per-redshift
# plot cells below. Set False to restore the old behaviour. Deliberately distinct
# from the composite figure's EXCLUDE_ZERO_NBR so later reassignments of that
# flag cannot leak into these cells.
EXCLUDE_ZERO_NBR_PLOTS = True
```

## Load Embeddings, Progenitors, and Properties

```python
# Load all 6 FIDUCIAL invdist embeddings (full-sample "_all" files: every galaxy
# passing the star-particle cut). Files are already N x K (K = n_neighbors_max),
# so no load-time truncation is needed.
# 'sorted_profiles' holds min(1/d, cap), right-aligned with the NEAREST neighbour
# in the LAST column; 'sorted_distances' holds the true sorted distances (zeros
# first, then ascending). Entries > 0 mark real neighbours in both. The stored
# (profile_mean, components) give EXACT linear projection into each snapshot's
# own PCA frame; the cross-snapshot projection below uses z=0's frame.
emb = {}
for s in SNAPS:
    d = np.load(f'snapshot{s}_invdist_r2_all.npz')
    prof = d['sorted_profiles']
    n_nbr = (prof > 0).sum(axis=1)
    emb[s] = {
        'embedding': d['embedding'],
        'original_index': d['original_index'].astype(int),
        'sorted_profiles': prof,
        'sorted_distances': d['sorted_distances'],
        'n_neighbors': n_nbr,
        'n_neighbors_max': int(d['n_neighbors_max']),
        'profile_mean': d['profile_mean'],
        'components': d['components'],
        'cap': float(d['cap']),
    }
    print(f'snap {s}: {emb[s]["embedding"].shape[0]} galaxies, '
          f'max n_neighbors = {emb[s]["n_neighbors_max"]}, K = {prof.shape[1]}')

# Load progenitor matches and properties (full-sample versions, targets 50/33/17)
prog = pd.read_csv('progenitors_multi_all.csv')
props = pd.read_csv('progenitor_properties_all.csv')
print(f'progenitors_multi_all.csv: {len(prog)} rows')
print(f'progenitor_properties_all.csv: {len(props)} rows')
```

```python
# Build SubfindID -> embedding-row lookup tables per snapshot
sid_to_row = {s: {int(sid): i for i, sid in enumerate(emb[s]['original_index'])} for s in SNAPS}

# For each z=0 galaxy, find its embedding-row at every snapshot (or -1 if not present)
n_z0 = len(prog)
row_at = {s: np.full(n_z0, -1, dtype=int) for s in SNAPS}
for i, r in prog.iterrows():
    row_at[99][i] = sid_to_row[99].get(int(r['subfind_z0']), -1)
    for s in [50, 33, 25, 21, 17]:
        col = f'subfind_snap_{s}'
        if pd.notna(r[col]):
            row_at[s][i] = sid_to_row[s].get(int(r[col]), -1)

# Per z=0 galaxy: does its progenitor at s exist AND have >= 1 neighbor there?
# (zero-neighbor progenitors are the degenerate single-point class)
has_nbr_at = {}
for s in SNAPS:
    ok = row_at[s] >= 0
    hn = np.zeros(n_z0, dtype=bool)
    hn[ok] = emb[s]['n_neighbors'][row_at[s][ok]] > 0
    has_nbr_at[s] = hn

# Mask: z=0 galaxies that have a progenitor present in the embedding at every snapshot
all_present = np.ones(n_z0, dtype=bool)
for s in SNAPS:
    all_present &= (row_at[s] >= 0)
print(f'Galaxies with progenitors in all 6 embeddings: {all_present.sum()} / {n_z0}')
```

## Exact Projection of Earlier Snapshots into z=0's PCA Frame

With the squared kernel the embedding is PCA of the profile matrix, so projecting any profile into z=0's coordinates is exact and linear: `X = (profile − profile_mean_z0) @ components_z0.T`. No similarity matrices, no eigenproblem, no Nyström approximation. Narrower snapshots are **left-padded** with zeros to z=0's width (profiles are right-aligned; a leading zero means "neighbour at infinity", exactly).

**Open design question** (CLAUDE.md pivot to-do 2): high-z profiles are uniformly lower (global density-scale evolution); this notebook projects them as-is, so part of any trajectory drift is that global scale change.


```python
# Align all snapshots to z=0's profile width by LEFT-PADDING with zeros
# (profiles are right-aligned, so leading zeros = absent neighbours = 1/inf,
# exactly). NEVER truncate below a snapshot's own K -- that would discard real
# neighbours. z=0 is the widest here; assert rather than assume.
K_ref = emb[99]['sorted_profiles'].shape[1]
max_K = max(emb[s]['sorted_profiles'].shape[1] for s in SNAPS)
assert K_ref == max_K, (
    f"z=0 width {K_ref} is not the widest ({max_K}); left-pad to the max instead")
sn = {}
for s in SNAPS:
    p = emb[s]['sorted_profiles']
    if p.shape[1] < K_ref:
        p = np.hstack([np.zeros((p.shape[0], K_ref - p.shape[1])), p])
    sn[s] = p
    print(f"snap {s}: profiles {emb[s]['sorted_profiles'].shape} -> padded {sn[s].shape}")
```

```python
# z=0's stored PCA frame IS the reference coordinate system -- no eigenproblem.
# (profile_mean, components) come from the SVD in generate_embedding_invdist.py
# and reproduce the stored embedding exactly:
profile_mean_z0 = emb[99]['profile_mean']
components_z0  = emb[99]['components']
X_z0 = (sn[99] - profile_mean_z0) @ components_z0.T
err = np.abs(X_z0 - emb[99]['embedding']).max()
print(f'Max diff from stored z=0 embedding: {err:.2e}')
assert err < 1e-8, 'stored frame does not reproduce the stored embedding'
```

```python
components_z0.shape
```

```python
# Project every snapshot's (left-padded) profiles into z=0's frame.
# Exact and linear -- replaces the legacy raw-kernel Nystrom machinery.
def project_to_z0(profiles_padded):
    return (profiles_padded - profile_mean_z0) @ components_z0.T

X = {99: X_z0}
for s in [50, 33, 25, 21, 17]:
    X[s] = project_to_z0(sn[s])
    print(f'snap {s} projected: {X[s].shape}')
```

```python
# Re-project z=0 through its OWN frame: must reproduce the stored embedding
X99_reproj = project_to_z0(sn[99])
X99_ref = emb[99]['embedding']

diff = np.abs(X99_reproj - X99_ref)
print(f"max  abs difference: {diff.max():.3e}")
print(f"mean abs difference: {diff.mean():.3e}")
print(f"allclose (atol=1e-8): {np.allclose(X99_reproj, X99_ref, atol=1e-8)}\n")

for d in range(X99_ref.shape[1]):
    r = np.corrcoef(X99_reproj[:, d], X99_ref[:, d])[0, 1]
    print(f"Dim {d+1}: corr={r:+.6f}, max|Δ|={diff[:, d].max():.3e}")
```

```python
# With the squared kernel the frame is exact PCA, so "variance explained" is
# legitimate. Per-dim shares of the total z=0 profile variance:
total_var = ((sn[99] - profile_mean_z0) ** 2).sum()
ev = (X_z0 ** 2).sum(axis=0)
cum = 0.0
for d in range(X_z0.shape[1]):
    cum += ev[d] / total_var
    print(f'Dim {d+1}: share {ev[d]/total_var:.3f}   cumulative {cum:.3f}')
```

```python
# Build per-galaxy trajectories: shape (n_tracked, n_snaps, n_dims)
tracked = np.where(all_present)[0]
snaps_ordered = sorted(SNAPS)  # 17, 21, 25, 33, 50, 99
zs = np.array([REDSHIFTS[s] for s in snaps_ordered])

n_dims = X[99].shape[1]                       # <-- derive from the embedding, don't reuse k
traj = np.empty((len(tracked), len(snaps_ordered), n_dims))
for j, s in enumerate(snaps_ordered):
    rows = row_at[s][tracked]
    traj[:, j, :] = X[s][rows]
if EXCLUDE_ZERO_NBR_PLOTS:
    # Zero-neighbor progenitors all sit at one exact projected point; at z=5 they
    # are >50% of `tracked` and would pin the plotted medians by arithmetic.
    for j, s in enumerate(snaps_ordered):
        zero = emb[s]['n_neighbors'][row_at[s][tracked]] == 0
        traj[zero, j, :] = np.nan
        if zero.any():
            print(f'  masked {zero.sum():4d} zero-neighbor progenitors at z={REDSHIFTS[s]:g}')

print(f'Trajectory tensor: {traj.shape}  (galaxies, snapshots, dims)')
print(f'NOTE: requiring a progenitor in all 6 embeddings keeps {len(tracked)}/{n_z0} '
      f'({len(tracked)/n_z0:.1%}) of z=0 galaxies -- a strongly mass-biased subsample '
      f'(and the 238 tree-less cluster-core satellites are absent by construction).')
```

## Evolution Plot 2 — Stratified by z=0 Embedding Dim 1

Mean ± standard error of the mean, for Dims 1, 2, 3 separately, stratified into tertiles of z=0 Dim 1 (Clusters / Filaments / Field).

```python
def plot_stratified_evolution(traj, zs, strat_values, strat_label, n_strata=3, labels=None):
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    edges = np.nanpercentile(strat_values, np.linspace(0, 100, n_strata + 1))
    bin_idx = np.clip(np.digitize(strat_values, edges) - 1, 0, n_strata - 1)
    bin_idx[~np.isfinite(strat_values)] = -1      # masked galaxies join no stratum
    colors = plt.cm.viridis(np.linspace(0, 0.9, n_strata))
    if labels is None:
        labels = [f'{strat_label} $\\in$ [{edges[i]:.2f}, {edges[i+1]:.2f}]' for i in range(n_strata)]
    for dim_i, ax in enumerate(axes):
        for k_strat in range(n_strata):
            sel = bin_idx == k_strat
            vals = traj[sel, :, dim_i]
            n_ok = np.maximum(np.sum(np.isfinite(vals), axis=0), 1)
            mu = np.nanmean(vals, axis=0)
            sem = np.nanstd(vals, axis=0, ddof=1) / np.sqrt(n_ok)
            ax.plot(zs, mu, color=colors[k_strat], marker='o', lw=2, label=labels[k_strat])
            ax.fill_between(zs, mu - sem, mu + sem, color=colors[k_strat], alpha=0.25)
        ax.invert_xaxis()
        ax.set_xlabel('Redshift', fontsize=13)
        ax.set_ylabel(f'Embedding Dim {dim_i+1}', fontsize=13)
        ax.set_title(f'Dim {dim_i+1} (stratified by {strat_label})', fontsize=13)
        if dim_i == 0:
            ax.legend(fontsize=10, loc='best')
    plt.tight_layout()
    plt.show()

# Stratify by z=0 Dim 1
z0_dim1 = traj[:, -1, 0]  # last snap is z=0 (snap 99)
plot_stratified_evolution(traj, zs, z0_dim1, 'z=0 Dim 1', n_strata=3,
                           labels=['Clusters (low Dim 1)', 'Filaments (mid)', 'Field (high)'])
```

```python
# Stellar-to-mass ratio vs. redshift, comparing the extremes of z=0 Dim 1 within
# subplots stratified by z=0 halo mass (M_200crit of the parent FoF group).
# Sample: CENTRALS at z=0, with z=0 M_200 in [HALO_MASS_MIN, HALO_MASS_MAX].
#
# Summary statistic is the GEOMETRIC mean (mean of log10, then 10**).  The band is
# +/-1 standard DEVIATION in log space -- i.e. the galaxy-to-galaxy scatter within
# each cell, not the uncertainty on the mean.  Computing it in log space keeps the
# band symmetric on the log axis and strictly positive.

HALO_MASS_MIN = 1e11     # M_sun, M_200crit at z=0
HALO_MASS_MAX = 1e13
DENOM = 'mass_total'     # ratio denominator: 'mass_total' (SubhaloMass) or 'm200'
DIM1_PCT = 10            # compare the bottom and top this-many percent of Dim 1

snaps_ordered = sorted(SNAPS)  # [17, 21, 25, 33, 50, 99]
zs = np.array([REDSHIFTS[s] for s in snaps_ordered])

props_idx = props.set_index('idx_z0') if 'idx_z0' in props.columns else props


def _col(idx, name):
    return pd.to_numeric(props_idx.loc[idx, name], errors='coerce').to_numpy(dtype=float)


# --- sample selection: centrals at z=0, within the halo-mass window -----------
is_cen_z0 = _col(tracked, 'is_central_snap_99') == 1
m200_z0   = _col(tracked, 'm200_snap_99') * 1e10 / H                # -> M_sun
in_mass   = (np.isfinite(m200_z0) & (m200_z0 >= HALO_MASS_MIN)
             & (m200_z0 <= HALO_MASS_MAX))

sel_gal   = np.where(is_cen_z0 & in_mass)[0]     # positions within `tracked` / `traj`
tracked_c = np.asarray(tracked)[sel_gal]

print(f'tracked galaxies:            {len(tracked)}')
print(f'  centrals at z=0:           {int(is_cen_z0.sum())}')
print(f'  M_200(z=0) in window:      {int(in_mass.sum())}')
print(f'  both (final sample):       {len(tracked_c)}')
print(f'  window: {HALO_MASS_MIN:.0e} - {HALO_MASS_MAX:.0e} M_sun')

n = len(tracked_c)
mstar = np.full((n, len(snaps_ordered)), np.nan)
mden  = np.full((n, len(snaps_ordered)), np.nan)    # ratio denominator
m200  = np.full((n, len(snaps_ordered)), np.nan)    # halo mass, for stratification
for j, s in enumerate(snaps_ordered):
    mstar[:, j] = _col(tracked_c, f'mass_stars_snap_{s}')
    mden[:, j]  = _col(tracked_c, f'{DENOM}_snap_{s}')
    m200[:, j]  = _col(tracked_c, f'm200_snap_{s}')

with np.errstate(divide='ignore', invalid='ignore'):
    ratio = mstar / mden
ratio[~np.isfinite(ratio) | (mden <= 0)] = np.nan

# A geometric mean cannot use ratio == 0 (progenitors with no stars yet), so report
# how many are dropped.
with np.errstate(invalid='ignore'):
    n_zero = np.sum(ratio == 0, axis=0)
    n_ok = np.sum(ratio > 0, axis=0)
print('\nzero-stellar-mass progenitors excluded from the geometric mean:')
for j, s in enumerate(snaps_ordered):
    print(f'  z = {REDSHIFTS[s]:g}: {int(n_zero[j])} zero, {int(n_ok[j])} usable')

# z=0 stratifiers (same galaxy subset)
z0_dim1 = traj[sel_gal, -1, 0]
z0_m200 = m200[:, -1]

# --- Dim 1 extremes: bottom and top DIM1_PCT per cent -------------------------
d_lo, d_hi = np.nanpercentile(z0_dim1, [DIM1_PCT, 100 - DIM1_PCT])
dim1_groups = [
    (f'Bottom {DIM1_PCT}% (low Dim 1, dense)',  z0_dim1 <= d_lo, plt.cm.viridis(0.15)),
    (f'Top {DIM1_PCT}% (high Dim 1, sparse)',   z0_dim1 >= d_hi, plt.cm.viridis(0.85)),
]
print(f'\nDim 1 range [{np.nanmin(z0_dim1):.2f}, {np.nanmax(z0_dim1):.2f}]; '
      f'cuts at {d_lo:.3f} and {d_hi:.3f}')

# Outer stratification: z=0 halo mass quantiles
log_mh = np.log10(np.where(z0_m200 > 0, z0_m200, np.nan))
n_mh = 4
valid_mh = ~np.isnan(log_mh)
mh_edges = np.nanpercentile(log_mh, np.linspace(0, 100, n_mh + 1))
mh_idx = np.clip(np.digitize(log_mh, mh_edges) - 1, 0, n_mh - 1)
mh_idx[~valid_mh] = -1
mh_edges_msun_log = mh_edges + np.log10(1e10 / H)
mh_titles = [f'log$_{{10}}$(M$_{{200,z=0}}$/M$_\\odot$) $\\in$ '
             f'[{mh_edges_msun_log[i]:.2f}, {mh_edges_msun_log[i+1]:.2f}]'
             for i in range(n_mh)]

# Occupancy matrix -- makes sparse/empty cells visible before you read the plot
print('\ncell counts (rows = mass bin, cols = Dim 1 group):')
print(f"{'massbin':>8}" + ''.join(f'{lab.split()[0] + lab.split()[1]:>14}'
                                  for lab, _, _ in dim1_groups))
for k in range(n_mh):
    row = [int(((mh_idx == k) & g).sum()) for _, g, _ in dim1_groups]
    print(f'{k:>8}' + ''.join(f'{v:>14}' for v in row))

MIN_PER_CELL = 5              # don't draw a line from a handful of galaxies
denom_label = ('Total Subhalo Mass' if DENOM == 'mass_total' else 'M$_{200}$')

fig, axes = plt.subplots(1, n_mh, figsize=(6 * n_mh, 5), sharey=True)
axes = np.atleast_1d(axes)
for k, ax in enumerate(axes):
    in_mh_bin = (mh_idx == k)
    for label, grp, colour in dim1_groups:
        sel = in_mh_bin & grp
        if sel.sum() < MIN_PER_CELL:
            continue
        with np.errstate(divide='ignore', invalid='ignore'):
            logr = np.log10(np.where(ratio[sel, :] > 0, ratio[sel, :], np.nan))
        counts = np.sum(np.isfinite(logr), axis=0)

        mean_log = np.full(len(snaps_ordered), np.nan)
        with np.errstate(invalid='ignore'):
            mean_log[counts > 0] = np.nanmean(logr[:, counts > 0], axis=0)
            std_log = np.where(counts > 1,
                               np.nanstd(logr, axis=0, ddof=1),
                               np.nan)          # scatter, NOT divided by sqrt(n)

        centre = 10 ** mean_log                       # geometric mean
        lo, hi = 10 ** (mean_log - std_log), 10 ** (mean_log + std_log)

        ax.plot(zs, centre, color=colour, marker='o', lw=2,
                label=f'{label} (n={sel.sum()})')
        ax.fill_between(zs, lo, hi, color=colour, alpha=0.25)
    ax.invert_xaxis()
    ax.set_xlabel('Redshift', fontsize=13)
    ax.set_title(mh_titles[k], fontsize=11)
    ax.set_yscale('log')
    if ax.get_legend_handles_labels()[0]:
        ax.legend(fontsize=10)
axes[0].set_ylabel(f'Stellar Mass / {denom_label}  (geometric mean)', fontsize=13)
fig.suptitle(f'Stellar-to-{denom_label} Ratio vs. redshift  —  z=0 centrals, '
             f'M$_{{200}}$(z=0) = {HALO_MASS_MIN:.0e}–{HALO_MASS_MAX:.0e} M$_\\odot$\n'
             f'Subplots: z=0 M$_{{200}}$ quantiles. Lines: bottom/top {DIM1_PCT}% of z=0 Dim 1. '
             f'Bands: $\\pm$1 s.d. (galaxy scatter) in log space. '
             f'Cells with < {MIN_PER_CELL} galaxies omitted.', fontsize=13)
plt.tight_layout(rect=[0, 0, 1, 0.93])
plt.show()
```

## Embedding–Property Relationships

At each snapshot, plot the *projected* embedding dimensions vs. the physical property at the same snapshot. Each property gets one figure: rows = Dims 1/2/3, columns = snapshots (z=5, 4, 3, 2, 1, 0). Scatter is faint; red line is the running median.

```python
def running_median(x, y, n_bins=15, min_count=5):
    if len(x) == 0:
        return np.array([]), np.array([])
    edges = np.linspace(x.min(), x.max(), n_bins + 1)
    centers = 0.5 * (edges[:-1] + edges[1:])
    med = np.full(n_bins, np.nan)
    for i in range(n_bins):
        m = (x >= edges[i]) & (x < edges[i+1])
        if m.sum() >= min_count:
            med[i] = np.median(y[m])
    ok = ~np.isnan(med)
    return centers[ok], med[ok]

# Property descriptors: (column-prefix, axis-label, transform)
def log10_safe(arr):
    return np.log10(np.clip(arr, 1e-30, None))

PROPS = [
    ('sfr',         'log$_{10}$(SFR / [M$_\\odot$/yr]), SF only',   log10_safe),
    ('mass_stars',  'log$_{10}$(M$_*$/M$_\\odot$)',                lambda a: log10_safe(a * 1e10 / H)),
    ('mass_total',  'log$_{10}$(M$_{tot}$/M$_\\odot$)',            lambda a: log10_safe(a * 1e10 / H)),
    ('m200',        'log$_{10}$(M$_{200}$/M$_\\odot$)',            lambda a: log10_safe(a * 1e10 / H)),
    ('starmet',     'Stellar metallicity (mass fraction)',         lambda a: a),
    ('gasmet',      'Gas metallicity (mass fraction)',             lambda a: a),
]
```

```python
from scipy.stats import pearsonr, spearmanr

for prefix, ylabel, transform in PROPS:
    fig, axes = plt.subplots(3, len(snaps_ordered),
                             figsize=(5 * len(snaps_ordered), 12), sharey='row')
    for col, s in enumerate(snaps_ordered):
        valid = row_at[s] >= 0
        if EXCLUDE_ZERO_NBR_PLOTS:
            valid &= has_nbr_at[s]
        col_name = f'{prefix}_snap_{s}'
        if col_name not in props.columns:
            for row in range(3):
                axes[row, col].set_visible(False)
            continue
        prop_vals = props.loc[valid, col_name].to_numpy(dtype=float)
        keep = ~np.isnan(prop_vals) & np.isfinite(prop_vals)
        if prefix == 'sfr':
            # star-forming only (sSFR >= 1e-11 /yr): SFR = 0 galaxies (22% at z=0)
            # would otherwise map to log10 = -30 and dominate the correlations
            ms = props.loc[valid, f'mass_stars_snap_{s}'].to_numpy(dtype=float) * 1e10 / H
            with np.errstate(divide='ignore', invalid='ignore'):
                ssfr = np.where(ms > 0, prop_vals / ms, np.nan)
            keep &= np.isfinite(ssfr) & (ssfr >= 1e-11)
        prop_t = transform(prop_vals[keep])
        idx_z0_valid = np.where(valid)[0][keep]
        emb_rows = row_at[s][idx_z0_valid]
        for row in range(3):
            ax = axes[row, col]
            x = X[s][emb_rows, row]
            # Drop any NaNs that snuck through the transform (e.g., log of 0)
            finite = np.isfinite(x) & np.isfinite(prop_t)
            xf, yf = x[finite], prop_t[finite]

            ax.scatter(xf, yf, s=4, alpha=0.15, color='steelblue', edgecolors='none')
            cx, cy = running_median(xf, yf)
            if len(cx):
                ax.plot(cx, cy, color='red', lw=2.5, marker='o', markersize=4)

            # Correlation coefficients
            if len(xf) > 2:
                r_p, _ = pearsonr(xf, yf)
                r_s, _ = spearmanr(xf, yf)
                ax.text(0.03, 0.97,
                        f'Pearson r = {r_p:+.2f}\nSpearman ρ = {r_s:+.2f}\nN = {len(xf)}',
                        transform=ax.transAxes, fontsize=9,
                        verticalalignment='top',
                        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.7))

            if row == 0:
                ax.set_title(f'z = {REDSHIFTS[s]}', fontsize=13)
            if col == 0:
                ax.set_ylabel(ylabel, fontsize=11)
            ax.set_xlabel(f'Embedding Dim {row+1} (projected, z=0 frame)', fontsize=10)

    fig.suptitle(f'{ylabel}  vs. embedding dimensions across redshift', fontsize=14)
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()
```

```python
from scipy.stats import binned_statistic

h = 0.6774          # TNG Hubble parameter (M200 code units are 1e10 Msun/h)
n_bins = 10         # bins for the satellite-fraction row
z_of_snap = {99: 0.0, 50: 1.0, 33: 2.0, 25: 3.0, 21: 4.0, 17: 5.0}
BAND_ALPHA = 0.18   # shading for the +/- 1 sigma bands


def running_median_std(x, y, n_bins=15, min_count=5):
    """Same binning as running_median(), but also returns the per-bin standard
    deviation of y, so the line can be drawn with a +/- 1 sigma band."""
    if len(x) == 0:
        return np.array([]), np.array([]), np.array([])
    edges = np.linspace(x.min(), x.max(), n_bins + 1)
    centers = 0.5 * (edges[:-1] + edges[1:])
    med = np.full(n_bins, np.nan)
    sd = np.full(n_bins, np.nan)
    for i in range(n_bins):
        m = (x >= edges[i]) & (x < edges[i+1])
        if m.sum() >= min_count:
            med[i] = np.median(y[m])
            sd[i] = np.std(y[m])
    ok = ~np.isnan(med)
    return centers[ok], med[ok], sd[ok]


# (left-axis label, kind) for each row
rows = [
    ('Star Formation Rate (M$_\\odot$/yr)\n(star-forming only)', 'sfr_linear'),
    ('log$_{10}$(M$_{200}$ / M$_\\odot$)',  'm200_log'),
    ('Satellite fraction',                   'satfrac'),
]

fig, axes = plt.subplots(3, 3, figsize=(18, 15), sharex='col', sharey='row')

for r, (ylabel, kind) in enumerate(rows):
    for dim in range(3):
        ax = axes[r, dim]
        for s in snaps_ordered:
            valid = row_at[s] >= 0
            if EXCLUDE_ZERO_NBR_PLOTS:
                valid &= has_nbr_at[s]
            label = f'z = {z_of_snap[s]:g}' if s in z_of_snap else f'snap {s}'

            if kind == 'satfrac':
                idx_z0_valid = np.where(valid)[0]
                emb_rows = row_at[s][idx_z0_valid]
                x = X[s][emb_rows, dim]
                y = (props.loc[idx_z0_valid, f'is_central_snap_{s}'] == 0).to_numpy().astype(float)
                edges = np.linspace(x.min(), x.max(), n_bins + 1)
                centers = 0.5 * (edges[:-1] + edges[1:])
                frac, _, _ = binned_statistic(x, y, statistic='mean', bins=edges)
                sd, _, _ = binned_statistic(x, y, statistic='std', bins=edges)
                counts, _ = np.histogram(x, bins=edges)
                ok = counts >= 10          # MIN_PER_BIN, matching the composite figure
                line, = ax.plot(centers[ok], frac[ok], lw=2.5, marker='o', markersize=4,
                                color=zcolor(REDSHIFTS[s]), label=label)
                # satellite fraction is a probability, so clip the band to [0, 1]
                lo = np.clip(frac[ok] - sd[ok], 0.0, 1.0)
                hi = np.clip(frac[ok] + sd[ok], 0.0, 1.0)
                ax.fill_between(centers[ok], lo, hi,
                                color=line.get_color(), alpha=BAND_ALPHA, lw=0)

            else:  # continuous property -> running median
                base = 'sfr' if kind == 'sfr_linear' else 'm200'
                col = f'{base}_snap_{s}'
                if col not in props.columns:
                    continue
                vals = props.loc[valid, col].to_numpy(dtype=float)
                keep = np.isfinite(vals) & ~np.isnan(vals)
                if kind == 'sfr_linear':
                    # star-forming only (sSFR >= 1e-11 /yr)
                    ms = props.loc[valid, f'mass_stars_snap_{s}'].to_numpy(dtype=float) * 1e10 / h
                    with np.errstate(divide='ignore', invalid='ignore'):
                        ssfr = np.where(ms > 0, vals / ms, np.nan)
                    keep &= np.isfinite(ssfr) & (ssfr >= 1e-11)
                if kind == 'm200_log':
                    keep &= vals > 0
                    y = np.log10(vals[keep] * 1e10 / h)
                else:
                    y = vals[keep]                       # linear SFR
                idx_z0_valid = np.where(valid)[0][keep]
                emb_rows = row_at[s][idx_z0_valid]
                x = X[s][emb_rows, dim]
                cx, cy, csd = running_median_std(x, y)
                if len(cx):
                    line, = ax.plot(cx, cy, lw=2.5, marker='o', markersize=4,
                                    color=zcolor(REDSHIFTS[s]), label=label)
                    lo, hi = cy - csd, cy + csd
                    if kind == 'sfr_linear':
                        lo = np.clip(lo, 0.0, None)      # SFR cannot be negative
                    ax.fill_between(cx, lo, hi,
                                    color=line.get_color(), alpha=BAND_ALPHA, lw=0)

        if kind == 'satfrac':
            ax.set_ylim(0, 1)
        if kind == 'sfr_linear':
            ax.set_ylim(0, 15)
        if r == len(rows) - 1:
            ax.set_xlabel(f'Embedding Dim {dim + 1}', fontsize=20)
        if dim == 0:
            ax.set_ylabel(ylabel, fontsize=20)
        ax.tick_params(labelsize=16)

# legend in the top-left subplot
axes[0, 0].legend(fontsize=16, loc='best')

plt.tight_layout()
plt.show()
```

```python
from scipy.stats import binned_statistic

# ----------------------------- config -----------------------------
SNAPS_SHOW  = sorted(SNAPS)    # all six; use [99, 33, 17] for a 3-curve version
N_BINS      = 10               # equal-width bins -> evenly spaced along the axis
MIN_PER_BIN = 10               # matters a lot with equal-width bins: controls tail extent
XCLIP       = (1, 99)          # percentile range for shared x-limits
SHOW_BANDS  = 'all'            # 'all' | 'endpoints' | 'none'
BAND_KIND   = 'se'             # 'se'         -> standard error on the plotted statistic
                               #                 (uncertainty; comparable across rows)
                               # 'sd'         -> +/-1 sample standard deviation of the
                               #                 galaxies in each bin (scatter)
                               # 'percentile' -> 16-84th percentile spread; median rows
                               #                 only, fraction rows fall back to 'sd'
EXCLUDE_ZERO_NBR = True
SSFR_CUT     = 1e-11           # quenched threshold, 1/yr
SFR_RELATIVE = False           # True -> divide by each redshift's median SFR
H = 0.6774

# (row label, kind, fixed ylim or None, line statistic)
ROWS = [
    ('Quenched fraction',                                'quenched',  (0, 1), 'mean'),
    ('SFR (M$_\\odot$/yr)\n(star-forming only)',         'sfr',       None,   'median'),
    ('log$_{10}$(M$_{200}$ / M$_\\odot$)',               'm200',      None,   'median'),
    ('Satellite fraction',                               'satellite', (0, 1), 'mean'),
]

props_i = props.set_index('idx_z0') if 'idx_z0' in props.columns else props
snaps_show = sorted(SNAPS_SHOW, key=lambda s: -REDSHIFTS[s])   # high z first
zvals = [REDSHIFTS[s] for s in snaps_show]
zmin, zmax = min(zvals), max(zvals)

# Fixed colour per redshift -- sequential blue->red ramp, uniform in CIELAB lightness.
# Keyed by redshift rather than by position, so a given z keeps its colour whether you
# plot all six snapshots or a subset.
ZCOLORS = {
    5.0: '#ff8f86',   # pale salmon
    4.0: '#ff5650',
    3.0: '#ea082a',   # red
    2.0: '#6c22e8',
    1.0: '#0202da',   # blue
    0.0: '#06068d',   # dark navy
}


def zcolor(z):
    """Colour for a redshift. Falls back to the nearest tabulated z, so exact catalog
    values (0.9973, 3.0081, ...) work as well as the rounded ones."""
    z = float(z)
    if z in ZCOLORS:
        return ZCOLORS[z]
    return ZCOLORS[min(ZCOLORS, key=lambda k: abs(k - z))]


def snap_data(s, kind, exclude_zero=EXCLUDE_ZERO_NBR):
    """(embedding rows, property values) for snapshot s, cleaned.

    exclude_zero is frozen to EXCLUDE_ZERO_NBR's value at DEFINITION time, so a
    later reassignment of that global elsewhere in the notebook cannot silently
    change this cell's figures on re-run. Pass exclude_zero=False explicitly to
    keep the zero-neighbour galaxies.
    """

    idx = np.where(row_at[s] >= 0)[0]
    rows = row_at[s][idx]

    def col(c):
        return pd.to_numeric(props_i.reindex(idx)[f'{c}_snap_{s}'],
                             errors='coerce').to_numpy(float)

    if kind == 'm200':
        v = col('m200')
        ok = np.isfinite(v) & (v > 0)
        y = np.log10(v[ok] * 1e10 / H)
    elif kind == 'satellite':
        v = col('is_central')
        ok = np.isfinite(v)
        y = (v[ok] == 0).astype(float)
    elif kind in ('quenched', 'sfr'):
        sfr_v, ms = col('sfr'), col('mass_stars') * 1e10 / H
        with np.errstate(divide='ignore', invalid='ignore'):
            ssfr = np.where(ms > 0, sfr_v / ms, np.nan)
        if kind == 'quenched':
            ok = np.isfinite(ssfr)
            y = (ssfr[ok] < SSFR_CUT).astype(float)
        else:
            # star-forming only -- the quenched row above accounts for the rest.
            # Including SFR = 0 galaxies would pin the median to zero wherever
            # more than half of a bin is quenched.
            ok = np.isfinite(ssfr) & (ssfr >= SSFR_CUT)
            y = sfr_v[ok]
            if SFR_RELATIVE:
                y = y / np.median(y)
    else:
        raise ValueError(kind)

    rows = rows[ok]
    if exclude_zero:
        keep = (emb[s]['sorted_profiles'] > 0).sum(axis=1)[rows] > 0
        rows, y = rows[keep], y[keep]
    return rows, y


def profile(x, y, stat):
    """Equal-width binned profile. Returns (x, y, lo, hi, counts) or None.

    Bins span each snapshot's own x-range, so every curve covers its full extent.
    Markers sit at the median x within each bin, not the bin centre.
    Band meaning is set by BAND_KIND (see config).
    """
    if len(x) < MIN_PER_BIN or x.max() <= x.min():
        return None
    edges = np.linspace(x.min(), x.max(), N_BINS + 1)
    cnt, _, _ = binned_statistic(x, y, 'count',  bins=edges)
    xm,  _, _ = binned_statistic(x, x, 'median', bins=edges)
    ym,  _, _ = binned_statistic(x, y, stat,     bins=edges)
    sd,  _, _ = binned_statistic(                              # ddof = 1 (sample s.d.)
        x, y, lambda v: np.std(v, ddof=1) if len(v) > 1 else np.nan, bins=edges)
    n = np.maximum(cnt, 1)

    with np.errstate(invalid='ignore', divide='ignore'):
        if stat == 'median':
            if BAND_KIND == 'sd':
                lo, hi = ym - sd, ym + sd
            elif BAND_KIND == 'percentile':
                lo, _, _ = binned_statistic(x, y, lambda v: np.percentile(v, 16), bins=edges)
                hi, _, _ = binned_statistic(x, y, lambda v: np.percentile(v, 84), bins=edges)
            else:                                              # 'se'
                p16, _, _ = binned_statistic(x, y, lambda v: np.percentile(v, 16), bins=edges)
                p84, _, _ = binned_statistic(x, y, lambda v: np.percentile(v, 84), bins=edges)
                sigma = 0.5 * (p84 - p16)                      # robust spread estimate
                se = 1.2533 * sigma / np.sqrt(n)               # standard error of the median
                lo, hi = ym - se, ym + se
        else:                                                  # fractions (mean of 0/1)
            if BAND_KIND in ('sd', 'percentile'):
                lo, hi = ym - sd, ym + sd
            else:                                              # 'se'
                se = np.sqrt(np.clip(ym * (1 - ym), 0, None) / n)
                lo, hi = ym - se, ym + se
            lo, hi = np.clip(lo, 0, 1), np.clip(hi, 0, 1)

    ok = cnt >= MIN_PER_BIN
    if not ok.any():
        return None
    return xm[ok], ym[ok], lo[ok], hi[ok], cnt[ok]


# sample sizes -- zero-neighbour galaxies are a systematic cut (they are the most
# isolated ones), and the excluded fraction rises steeply with redshift.
print(f"{'z':>5} {'matched':>9} {'zero-nbr':>9} {'excl %':>8} "
      f"{'used':>8} {'SF':>8} {'SF %':>7}")
for s in snaps_show:
    rows_raw, y_raw = snap_data(s, 'quenched', exclude_zero=False)
    n_zero = int(((emb[s]['sorted_profiles'] > 0).sum(axis=1)[rows_raw] == 0).sum())
    n_all = len(snap_data(s, 'quenched')[1])
    n_sf = len(snap_data(s, 'sfr')[1])
    print(f'{REDSHIFTS[s]:>5g} {len(y_raw):>9} {n_zero:>9} '
          f'{n_zero / max(len(y_raw), 1):>8.1%} {n_all:>8} {n_sf:>8} '
          f'{n_sf / max(n_all, 1):>7.1%}')

# --------------------------- the figure ---------------------------
fig, axes = plt.subplots(len(ROWS), 3, figsize=(16, 4.6 * len(ROWS)),
                         sharex='col', sharey='row')

xlims = {}
for dim in range(3):
    pooled = np.concatenate([X[s][snap_data(s, 'm200')[0], dim] for s in snaps_show])
    xlims[dim] = np.percentile(pooled, XCLIP)

for r, (ylabel, kind, ylim, stat) in enumerate(ROWS):
    row_vals = []
    for dim in range(3):
        ax = axes[r, dim]
        for s in snaps_show:
            rows, y = snap_data(s, kind)
            p = profile(X[s][rows, dim], y, stat)
            if p is None:
                continue
            xm, ym, lo, hi, cnt = p
            c = zcolor(REDSHIFTS[s])
            band = (SHOW_BANDS == 'all' or
                    (SHOW_BANDS == 'endpoints' and REDSHIFTS[s] in (zmin, zmax)))
            if band:
                ax.fill_between(xm, lo, hi, color=c, alpha=0.25, lw=0, zorder=1)
            ax.plot(xm, ym, color=c, lw=2, marker='o', markersize=4,
                    label=f'z = {REDSHIFTS[s]:g}', zorder=3)
            row_vals.append(ym)
        ax.set_xlim(*xlims[dim])
        if r == len(ROWS) - 1:
            ax.set_xlabel(f'Embedding Dim {dim + 1}', fontsize=19)
        if dim == 0:
            ax.set_ylabel(ylabel, fontsize=19)
        ax.tick_params(labelsize=15)

    # y-limits: fixed for fractions (padded so a 0.0 line clears the spine),
    # robust auto-zoom otherwise, so one outlying bin cannot set the row's scale
    if ylim is not None:
        axes[r, 0].set_ylim(ylim[0] - 0.03, ylim[1] + 0.03)
    elif row_vals:
        allv = np.concatenate(row_vals)
        lo_v, hi_v = np.percentile(allv, [1, 99])
        span = hi_v - lo_v
        pad = 0.08 * (span if span > 0 else 1.0)
        axes[r, 0].set_ylim(lo_v - pad, hi_v + pad)

axes[0, 0].legend(fontsize=16, frameon=False, loc='best')
plt.tight_layout()
plt.show()
```

```python
from scipy.stats import binned_statistic_2d

# ---------------------------------------------------------------- config
SNAP = 99            # snapshot to map; 99 = z=0
N_BINS = 30          # 2D bins per axis
MIN_COUNT = 10       # bins with fewer galaxies are left blank
CLIP_PCT = (0.5, 99.5)   # robust axis limits, drops extreme outliers
EXCLUDE_ZERO_NBR_2D = False  # this cell's own flag: 2D maps keep the degenerate point
h = 0.6774

DIM_PAIRS = [(0, 1), (0, 2), (1, 2)]   # (Dim1,Dim2), (Dim1,Dim3), (Dim2,Dim3)

# (row label, column prefix, kind, colormap, fixed color limits or None)
ROWS = [
    ('Median SFR (M$_\\odot$/yr)',              'sfr',        'linear',  'viridis', None),
    ('Median log$_{10}$(M$_{200}$/M$_\\odot$)', 'm200',       'log10',   'viridis', None),
    ('Satellite fraction',                       'is_central', 'satfrac', 'cividis', (0.0, 1.0)),
]

# props may or may not already be indexed by idx_z0 (cell 54 re-indexes it)
props_i = props.set_index('idx_z0') if 'idx_z0' in props.columns else props

# ---------------------------------------------------------------- data prep
valid = row_at[SNAP] >= 0
idx_all = np.where(valid)[0]
emb_all = row_at[SNAP][idx_all]

if EXCLUDE_ZERO_NBR_2D:
    n_nbr = (emb[SNAP]['sorted_profiles'] > 0).sum(axis=1)
    keep = n_nbr[emb_all] > 0
    idx_all, emb_all = idx_all[keep], emb_all[keep]
    print(f'excluded {(~keep).sum()} zero-neighbor galaxies')

coords = X[SNAP][emb_all]        # (N, n_dims) embedding coords, aligned to idx_all

# Shared axis ranges per dimension, so every panel in a column is comparable
lims = {d: np.percentile(coords[:, d], CLIP_PCT) for d in range(3)}

def property_values(prefix, kind):
    """Return (mask into idx_all, values) for one property row."""
    v = pd.to_numeric(props_i.loc[idx_all, f'{prefix}_snap_{SNAP}'],
                      errors='coerce').to_numpy(dtype=float)
    ok = np.isfinite(v)
    if kind == 'log10':
        ok &= v > 0
        return ok, np.log10(v[ok] * 1e10 / h)
    if kind == 'satfrac':
        return ok, (v[ok] == 0).astype(float)   # is_central == 0  ->  satellite
    return ok, v[ok]                            # linear

# ---------------------------------------------------------------- figure
fig, axes = plt.subplots(3, 3, figsize=(19, 16), sharex='col', sharey='col')

for r, (ylabel, prefix, kind, cmap_name, fixed_lims) in enumerate(ROWS):
    ok, vals = property_values(prefix, kind)
    xy = coords[ok]

    # Pass 1: bin all three panels before choosing a shared color scale
    panels = []
    for (dx, dy) in DIM_PAIRS:
        xedges = np.linspace(*lims[dx], N_BINS + 1)
        yedges = np.linspace(*lims[dy], N_BINS + 1)
        stat = 'mean' if kind == 'satfrac' else 'median'
        med, _, _, _ = binned_statistic_2d(xy[:, dx], xy[:, dy], vals,
                                           statistic=stat, bins=[xedges, yedges])
        cnt, _, _, _ = binned_statistic_2d(xy[:, dx], xy[:, dy], vals,
                                           statistic='count', bins=[xedges, yedges])
        M = np.ma.masked_invalid(np.where(cnt >= MIN_COUNT, med, np.nan))
        panels.append((xedges, yedges, M))

    # One color scale for the whole row, robust to a few extreme bins
    if fixed_lims is not None:
        vmin, vmax = fixed_lims
    else:
        pooled = np.concatenate([M.compressed() for _, _, M in panels])
        vmin, vmax = np.percentile(pooled, [2, 98])

    cmap = plt.get_cmap(cmap_name).copy()
    cmap.set_bad(alpha=0.0)          # empty / sparse bins show the white page

    # Pass 2: draw
    for c, ((dx, dy), (xedges, yedges, M)) in enumerate(zip(DIM_PAIRS, panels)):
        ax = axes[r, c]
        ax.set_facecolor('white')
        mesh = ax.pcolormesh(xedges, yedges, M.T, cmap=cmap,
                             vmin=vmin, vmax=vmax, shading='flat', rasterized=True)
        if r == len(ROWS) - 1:
            ax.set_xlabel(f'Embedding Dim {dx + 1}', fontsize=20)
        if c == 0:
            ax.set_ylabel(f'Embedding Dim {dy + 1}', fontsize=20)
        else:
            ax.set_ylabel(f'Embedding Dim {dy + 1}', fontsize=20)
        ax.tick_params(labelsize=16)

    cbar = fig.colorbar(mesh, ax=axes[r, :].tolist(), fraction=0.025, pad=0.015)
    cbar.set_label(ylabel, fontsize=18)
    cbar.ax.tick_params(labelsize=14)

fig.suptitle(f'Median physical properties across the embedding plane  '
             f'(z = {REDSHIFTS[SNAP]:g}, bins with < {MIN_COUNT} galaxies blank)',
             fontsize=20, y=0.995)
plt.show()
```
